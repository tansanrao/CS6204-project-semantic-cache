# Local Proxy Quickstart

This notebook demonstrates how to send OpenAI-compatible requests to the local proxy running at `http://localhost:8080/v1`. Update the connection settings if the proxy is hosted elsewhere or if authentication is enabled.

```python
import os
from openai import OpenAI

# Configure connection details
PROXY_BASE_URL = os.getenv("PROXY_BASE_URL", "http://localhost:8080/v1")
PROXY_API_KEY = os.getenv("PROXY_API_KEY", "not-needed")
MODEL_NAME = os.getenv("PROXY_MODEL", "openai/gpt-oss-20b")

client = OpenAI(base_url=PROXY_BASE_URL, api_key=PROXY_API_KEY)
print(f"Using proxy at {PROXY_BASE_URL}")
print(f"Model: {MODEL_NAME}")
```

```python
# Optional: enumerate models exposed by the proxy
models = client.models.list()
print(f"Found {len(models.data)} models")
for model in models.data:
    print(model.id)
```

```python
# Send a test chat completion request
response = client.chat.completions.create(
    model=MODEL_NAME,
    messages=[
        {"role": "system", "content": "You are a concise assistant that explains TTL strategies clearly."},
        {"role": "user", "content": "Give me a one-sentence summary of how TTL caching helps APIs stay fresh."},
    ],
    max_tokens=128,
)

message = response.choices[0].message
print(message.role, ":", message.content.strip())
```
